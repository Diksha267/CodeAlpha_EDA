The dataset used for this project is Students Performance in Exams. It contains information about students, including their exam scores in mathematics, reading and writing.

## 2. Data Quality

The dataset was checked for missing values and duplicate records to ensure data quality.

Missing values: 0 in all columns

Duplicate rows: 0

## 3. Score Distribution

The distribution of math scores was analyzed using a histogram to understand the spread and frequency of students' scores.

## 4. Subject Comparison

The average scores in mathematics, reading and writing were compared using a bar chart to identify differences in student performance.

## 5. Correlation Analysis

A correlation heatmap was created to understand the relationships between mathematics, reading and writing scores.

## 6. Conclusion

Exploratory Data Analysis helped in understanding the dataset, checking data quality, identifying patterns and comparing students' performance across different subjects. Visualizations and statistical analysis were used to gain meaningful insights from the data.






```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/StudentsPerformance.csv')

print(df.head())
```

```python
# Number of rows and columns
print(df.shape)

# Columns Names
print(df.columns)

# Data Types and non - null counts
df.info()
```

```python
# Find missing values

print(df.isnull().sum())
```

```python
# Find duplicate rows

print("Duplicate rows:", df.duplicated().sum())
```

```python
# Get statistical information

print(df.describe())
```

```python
# Distribution of math scores

plt.figure(figsize=(8, 5))
sns.histplot(df['math score'], bins=15, kde=True)
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Number of Students')
plt.show()
```

```python
# Compare average scores

scores = ['math score', 'reading score', 'writing score']

df[scores].mean().plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Average Scores by Subject')
plt.ylabel('Average Score')
plt.xlabel('Subject')
plt.xticks(rotation=0)
plt.show()
```

```python
# Correlation between scores

plt.figure(figsize=(7, 5))

sns.heatmap(
    df[['math score', 'reading score', 'writing score']].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Between Scores')
plt.show()
```
